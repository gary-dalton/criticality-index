### List all "Manual Review" prefixes

```julia
using DataFrames, CSV

"""
Extracts unique prefixes from 'Manual Review' rows and saves them to a CSV.
"""
function export_unknown_prefixes(manifest_path::String, output_path::String; slug_col=:Variable)
    # 1. Load the manifest
    if !isfile(manifest_path)
        error("Manifest file not found at $manifest_path")
    end
    df = CSV.read(manifest_path, DataFrame)
    
    # 2. Identify 'Unknown' or 'Manual Review' rows
    # We look for rows where the categorization logic fell through
    unknown_df = filter(row -> 
        (ismissing(row.data_nature) || row.data_nature == "Unknown") || 
        (ismissing(row.primary_citation) || row.primary_citation == "Manual Review"), 
        df
    )
    
    if nrow(unknown_df) == 0
        println("All prefixes categorized. No CSV generated.")
        return
    end
    
    # 3. Extract unique prefixes and their count (to prioritize high-volume datasets)
    # We get everything before the first underscore
    prefix_list = [split(string(row[slug_col]), "_")[1] for row in eachrow(unknown_df)]
    
    # 4. Create a summary DataFrame
    summary_df = combine(groupby(DataFrame(prefix = prefix_list), :prefix), nrow => :variable_count)
    sort!(summary_df, :variable_count, rev=true)
    
    # 5. Save to CSV
    CSV.write(output_path, summary_df)
    
    println("Exported $(nrow(summary_df)) unique unknown prefixes to $output_path")
    return summary_df
end

# Execution
export_unknown_prefixes("work/data/qog_metadata_manifest_v2.csv", "work/data/unknown_prefixes.csv")
```

```julia
using DataFrames, Statistics, Plots, StatsPlots

# 1. Temporal Sparsity: When does the data "wake up"?
# We calculate the average density of all variables for each year.
temporal_audit = combine(groupby(df_std, :year)) do dd
    # Calculate % of non-missing values across all columns (excluding identifiers)
    data_cols = setdiff(names(dd), ["cname", "ccode", "year", "ccodealp", "ccode_qog"])
    total_cells = length(data_cols) * nrow(dd)
    non_missing = sum([sum(.!ismissing.(dd[!, col])) for col in data_cols])
    (Density = non_missing / total_cells,)
end

p1 = plot(temporal_audit.year, temporal_audit.Density, 
    fillrange=0, fillalpha=0.3, lw=3, color=:blue,
    title="Temporal Sparsity: Global Data Presence by Year",
    xlabel="Year", ylabel="Avg. Variable Density",
    legend=false)

# 2. Country Sparsity: Who are the "Invisible States"?
# We calculate which countries have the most "holes" in their record.
country_audit = combine(groupby(df_std, :cname)) do dd
    data_cols = setdiff(names(dd), ["cname", "ccode", "year", "ccodealp", "ccode_qog"])
    total_cells = length(data_cols) * nrow(dd)
    non_missing = sum([sum(.!ismissing.(dd[!, col])) for col in data_cols])
    (Density = non_missing / total_cells,)
end

sort!(country_audit, :Density)
# Showing the 20 most "Sparse" countries
p2 = bar(country_audit.cname[1:20], country_audit.Density[1:20],
    title="The 20 Most Sparse Countries (Data Voids)",
    xrotation=45, color=:red, alpha=0.7, legend=false,
    ylabel="Data Presence Ratio")

# 3. Display the Audit
display(plot(p1, p2, layout=(2,1), size=(900, 1000), margin=5Plots.mm))
```

```julia
using DataFrames, Statistics

# 1. Identify data columns (excluding base identifiers)
# We exclude the columns that are always present (ISO codes, Names, Year)
id_cols = ["cname", "ccode", "year", "ccodealp", "ccode_qog", "cname_year", "ccodealp_year"]
data_cols = setdiff(names(df_std), id_cols)

# 2. Compute Density per Country
# Density = (Total non-missing values) / (Total possible cells)
country_sparsity = combine(groupby(df_std, :cname)) do dd
    num_rows = nrow(dd)
    num_vars = length(data_cols)
    total_possible_cells = num_rows * num_vars
    
    # Sum up all non-missing entries across all data columns
    actual_data_points = sum([sum(.!ismissing.(dd[!, col])) for col in data_cols])
    
    (Density = round(actual_data_points / total_possible_cells, digits=4), 
     Total_Observations = actual_data_points)
end

# 3. Sort to find the "Top 20 Most Sparse" (Lowest Density first)
sort!(country_sparsity, :Density)

println("The 20 Most Sparse Countries/Areas (Data Voids):")
display(first(country_sparsity, 20))
```

## Establish slug metadata
With birth and death year from Sweden benchmark. Indcludes density in the benchmark (Sweden) during its life

```julia
using DataFrames, CSV, Arrow, Statistics

# 1. Load Data
manifest = CSV.read("work/data/qog_metadata_manifest.csv", DataFrame)
df_std = Arrow.Table("work/data/qog_std_ts_jan25.arrow") |> DataFrame

# Identify the correct column name from your printout above
# Replace :Variable with the correct symbol (e.g., :slug, :variable_name)
slug_col = :Variable 

# 2. Function for the Certificate
function get_slug_certificate(slug, data_df; benchmark="Sweden")
    # Safety check: does the slug exist in the dataframe?
    if !(string(slug) in names(data_df))
        return (missing, missing, 0.0)
    end
    
    bench_years = data_df[(data_df.cname .== benchmark) .& .!ismissing.(data_df[!, slug]), :year]
    
    if isempty(bench_years)
        return (missing, missing, 0.0)
    end
    
    birth, death = minimum(bench_years), maximum(bench_years)
    window_data = data_df[(data_df.cname .== benchmark) .& (data_df.year .>= birth) .& (data_df.year .<= death), slug]
    density = mean(.!ismissing.(window_data))
    
    return (birth, death, round(density, digits=4))
end

# 3. Update the Manifest
# We use 'string.' to ensure the slug is treated as a string column name
results = [get_slug_certificate(row[slug_col], df_std) for row in eachrow(manifest)]

manifest.birth_year = [r[1] for r in results]
manifest.death_year = [r[2] for r in results]
manifest.benchmark_density = [r[3] for r in results]

# 4. Finalize Step 1
CSV.write("work/data/qog_metadata_manifest_v2.csv", manifest)
println("Step 1 Complete. Check 'qog_metadata_manifest_v2.csv' for the new certificates.")
```

## Rerun sparsity analysis

```julia
using DataFrames, Statistics

# 1. Identify Data Columns
# We exclude identifiers and our new mask column
id_cols = ["cname", "ccode", "year", "ccodealp", "ccode_qog", "cname_year", "ccodealp_year", "is_valid"]
data_cols = setdiff(names(df_std), id_cols)

# 2. Calculate Density Relative to Lifespan
relative_sparsity = combine(groupby(df_std, :cname)) do dd
    # Only look at rows where the state was active/valid
    valid_rows = filter(row -> row.is_valid == true, dd)
    
    if nrow(valid_rows) == 0
        return (Relative_Density = 0.0, Valid_Years = 0, Total_Obs = 0)
    end
    
    # Calculate density across all variables for these years
    total_possible_cells = nrow(valid_rows) * length(data_cols)
    actual_data_points = sum([sum(.!ismissing.(valid_rows[!, col])) for col in data_cols])
    
    (Relative_Density = round(actual_data_points / total_possible_cells, digits=4),
     Valid_Years = nrow(valid_rows),
     Total_Obs = actual_data_points)
end

# 3. Sort to find the True Reporting Failures (Lowest Relative Density)
sort!(relative_sparsity, :Relative_Density)

println("Top 20 Most Sparse Countries (Adjusted for Active Sovereignty):")
display(first(relative_sparsity, 20))
```

## Rerun using comparison to Sweden

```julia
using DataFrames, Statistics, Arrow

# 1. Calculate the 'Gold Standard' Yearly Density (Sweden)
# We find the percentage of ALL slugs Sweden reports each year
sweden_data = filter(row -> row.cname == "Sweden", df_std)
sweden_yearly = combine(groupby(sweden_data, :year)) do dd
    d = mean([mean(.!ismissing.(dd[!, col])) for col in data_cols])
    (Benchmark_Density = d,)
end

# 2. Calculate the Gap for every Country
reporting_gap_stats = combine(groupby(df_std, :cname)) do dd
    # Only evaluate years where the country existed (is_valid == true)
    valid_rows = filter(row -> row.is_valid == true, dd)
    
    if nrow(valid_rows) == 0
        return (Gap = 0.0, Country_Density = 0.0, Benchmark_Avg = 0.0)
    end
    
    # Calculate this country's average density across its valid lifespan
    country_density = mean([mean(.!ismissing.(valid_rows[!, col])) for col in data_cols])
    
    # Get Sweden's benchmark for these exact same years
    comparison = leftjoin(valid_rows[:, [:year]], sweden_yearly, on = :year)
    benchmark_avg = mean(comparison.Benchmark_Density)
    
    (Gap = round(benchmark_avg - country_density, digits=4),
     Country_Density = round(country_density, digits=4),
     Benchmark_Avg = round(benchmark_avg, digits=4))
end

# 3. Sort by the largest Gap (True Reporting Failures)
sort!(reporting_gap_stats, :Gap, rev=true)

println("Top 20 Countries with the Largest Reporting Gap (vs Sweden's Era-Adjusted Performance):")
display(first(reporting_gap_stats, 20))
```

```julia
using DataFrames, Statistics

"""
    calculate_continuity(df, slug)
    
Calculates the average interval between observations. 
1.0 = Annual data. 
>1.0 = Gaps in the time series.
"""
function calculate_continuity(df::AbstractDataFrame, slug::String)
    # 1. Isolate the variable and remove missings
    # Ensure we use String for selection to match Method A
    data = sort(dropmissing(df[:, ["cname", "year", slug]]), ["cname", "year"])
    
    if isempty(data)
        return Inf
    end

    # 2. Calculate year-over-year differences per country
    # We use 'ungroup' behavior or flatten the results
    gaps_df = combine(groupby(data, "cname")) do dd
        if size(dd, 1) < 2
            return (gap = Int64[],) # Return empty typed vector for singletons
        else
            return (gap = diff(dd.year),)
        end
    end
    
    # 3. Handle the case where no countries have multiple years
    if isempty(gaps_df.gap)
        return Inf
    end
    
    # 4. Return the mean of all recorded year-gaps
    return mean(gaps_df.gap)
end

# Test with your study set
try
    score = calculate_continuity(df_study, "wbgi_rle")
    println("Continuity Score for wbgi_rle: ", round(score, digits=2))
catch e
    @error "Continuity calculation failed" exception=e
end
```

```julia
using DataFrames, Statistics

"""
    analyze_gap_structure(df, slug)
    
Calculates the length of internal gaps (missing years) in a time-series.
"""
function analyze_gap_structure(df::AbstractDataFrame, slug::String)
    # 1. Isolate and sort data
    # We use String selection to remain consistent with your 'Method A'
    data = sort(dropmissing(df[:, ["cname", "year", slug]]), ["cname", "year"])
    
    if isempty(data)
        return "Variable '$slug' has no data."
    end

    # 2. Calculate gap lengths per country
    # We find the difference between consecutive years and subtract 1
    # (e.g., 2020 - 2018 = 2. 2 - 1 = 1 year gap)
    gaps_df = combine(groupby(data, "cname"), :year => (x -> diff(x) .- 1) => :gap_list)
    
    # 3. Flatten the lists of gaps into a single vector of integers
    # Some countries might have multiple gaps, some none.
    all_gaps = reduce(vcat, gaps_df.gap_list, init=Int64[])
    
    # 4. Filter for actual gaps (where gap > 0)
    actual_gaps = filter(x -> x > 0, all_gaps)
    
    if isempty(actual_gaps)
        return "No internal gaps found in '$slug'. Data is perfectly continuous."
    end
    
    # 5. Statistical Profiling
    return (
        variable = slug,
        total_gaps = length(actual_gaps),
        max_gap_years = maximum(actual_gaps),
        mean_gap_years = round(mean(actual_gaps), digits=2),
        short_gaps_count = count(<=(2), actual_gaps),
        long_gaps_count = count(>(2), actual_gaps)
    )
end

# Usage
analyze_gap_structure(df_study, "wbgi_rle")
```

```julia
"""
    rank_gap_intensity(df, slug)
    
Returns a list of countries sorted by how frequently they skip data points.
"""
function rank_gap_intensity(df::AbstractDataFrame, slug::String)
    # Isolate data
    data = sort(dropmissing(df[:, ["cname", "year", slug]]), ["cname", "year"])
    
    # Calculate gaps per country
    gaps_per_country = combine(groupby(data, "cname")) do dd
        # Count occurrences where year-to-year jump is > 1
        n_gaps = count(diff(dd.year) .> 1)
        return (gap_count = n_gaps,)
    end
    
    # Return top 20 most 'gappy' countries
    return sort(gaps_per_country, :gap_count, rev=true)[1:20, :]
end

# Usage
rank_gap_intensity(df_study, "wbgi_rle")
```

```julia
using DataFrames, Statistics

"""
    analyze_multiple_slugs(df, slugs)
    
Aggregates gap diagnostics for a vector of variable names.
"""
function analyze_multiple_slugs(df::AbstractDataFrame, slugs::Vector{String})
    results = []
    
    for slug in slugs
        # Skip core identifiers
        if slug in ["cname", "year", "ccode"] continue end
        
        # Reuse our robust single-slug logic
        profile = analyze_gap_structure(df, slug)
        
        # If the result is a string (error/no data), skip or log
        if profile isa String
            continue
        end
        
        push!(results, profile)
    end
    
    return sort(DataFrame(results), :max_gap_years, rev=true)
end

# Usage: Analyze all your modern discovery slugs
gap_report = analyze_multiple_slugs(df_study, modern_slugs)
display(gap_report)
```
