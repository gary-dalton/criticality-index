## 2. QoG Alignment

```julia
qog = DataFrame(Arrow.Table("data/qog_std_ts_jan25_aug.arrow"))
qog_codes = Set(unique(skipmissing(qog.ident_ccodealp)))
shdi_codes = Set(unique(shdi.isocode3))

matched = length(intersect(shdi_codes, qog_codes))
in_shdi_not_qog = sort(collect(setdiff(shdi_codes, qog_codes)))
in_qog_not_shdi = sort(collect(setdiff(qog_codes, shdi_codes)))

println("QoG: $(length(qog_codes))  SHDI: $(length(shdi_codes))  Matched: $matched")
println()
println("In SHDI but NOT in QoG ($(length(in_shdi_not_qog))):")
for c in in_shdi_not_qog; println("  $c"); end
println()
println("In QoG but NOT in SHDI ($(length(in_qog_not_shdi))):")
for c in in_qog_not_shdi; println("  $c"); end
```

## 3. Missingness (Subnational)

```julia
sub = shdi_subnational(shdi)

pre = filter(r -> r.year < 2005, sub)
post = filter(r -> r.year >= 2005, sub)

println("Subnational missingness (sorted worst first):")
println("Column                  Pre-2005 (n=$(nrow(pre)))    Post-2005 (n=$(nrow(post)))    Overall (n=$(nrow(sub)))")
println("─"^90)
for col in names(sub)
    all_miss = round(100 * count(ismissing, sub[!, col]) / nrow(sub), digits=1)
    all_miss == 0 && continue
    pre_miss = nrow(pre) > 0 ? round(100 * count(ismissing, pre[!, col]) / nrow(pre), digits=1) : 0.0
    post_miss = nrow(post) > 0 ? round(100 * count(ismissing, post[!, col]) / nrow(post), digits=1) : 0.0
    println("  $(rpad(string(col), 22)) $(lpad(string(pre_miss), 5))% missing   $(lpad(string(post_miss), 5))% missing   $(lpad(string(all_miss), 5))% missing")
end
```

## 4. Coverage — Regions per Country

```julia
reg_per_country = combine(groupby(sub, [:isocode3, :country]),
    :gdlcode => (x -> length(unique(x))) => :n_regions,
    :year => minimum => :first_year,
    :year => maximum => :last_year
)
sort!(reg_per_country, :n_regions, rev=true)
println("Top 15 by region count:")
first(reg_per_country, 15)
```

```julia
println("Bottom 15 by region count:")
last(sort(reg_per_country, :n_regions, rev=true), 15)
```

## 5. Subnational Dispersion (CV of SHDI, 2020)

Compare to DOSE — SHDI covers 188 countries vs DOSE's 83.

```julia
using Statistics

yr2020 = filter(r -> r.year == 2020 && r.level == "Subnat" && !ismissing(r.shdi), sub)
dispersion = combine(groupby(yr2020, [:isocode3, :country]),
    :shdi => std => :sd_shdi,
    :shdi => mean => :mean_shdi,
    nrow => :n_regions
)
filter!(r -> r.n_regions >= 3, dispersion)
dispersion[!, :cv] = dispersion.sd_shdi ./ dispersion.mean_shdi
sort!(dispersion, :cv)

println("Top 15 most internally EQUAL (lowest CV of subnational HDI, 2020):")
first(select(dispersion, :isocode3, :country, :n_regions, :mean_shdi, :cv), 15)
```

```julia
println("Top 15 most internally UNEQUAL (highest CV):")
first(sort(select(dispersion, :isocode3, :country, :n_regions, :mean_shdi, :cv), :cv, rev=true), 15)
```

```julia
# Compare SHDI vs DOSE dispersion — countries in both datasets
dose_sub = DataFrame(Arrow.Table("data/dose_subnational.arrow"))
dose_2015 = filter(r -> r.year == 2015 && !ismissing(r.gdp_pc_usd2015), dose_sub)
dose_cv = combine(groupby(dose_2015, :iso3),
    :gdp_pc_usd2015 => std => :sd_gdp,
    :gdp_pc_usd2015 => mean => :mean_gdp,
    nrow => :n_regions_dose
)
filter!(r -> r.n_regions_dose >= 3, dose_cv)
dose_cv[!, :cv_gdp] = dose_cv.sd_gdp ./ dose_cv.mean_gdp

# SHDI 2020 CV (already computed as dispersion)
shdi_cv = select(dispersion, :isocode3 => :iso3, :cv => :cv_shdi, :n_regions => :n_regions_shdi)

# Join
comparison = innerjoin(dose_cv, shdi_cv, on=:iso3)
println("Countries in both DOSE and SHDI with ≥3 regions: $(nrow(comparison))")
println()
println("Correlation of GDP CV vs HDI CV: $(round(cor(comparison.cv_gdp, comparison.cv_shdi), digits=3))")
println()
sort!(comparison, :cv_shdi)
println("Comparison (sorted by SHDI CV):")
first(select(comparison, :iso3, :cv_gdp, :cv_shdi, :n_regions_dose, :n_regions_shdi), 20)
```

## 6. Gender Disaggregation

SHDI includes male/female splits for all indicators — unique among our datasets.

```julia
# Gender gap in SHDI (2020, subnational)
yr2020g = filter(r -> r.year == 2020 && r.level == "Subnat" && !ismissing(r.shdif) && !ismissing(r.shdim), sub)
yr2020g[!, :gender_gap] = yr2020g.shdif .- yr2020g.shdim

country_gap = combine(groupby(yr2020g, [:isocode3, :country]),
    :gender_gap => mean => :mean_gap,
    :gender_gap => std => :sd_gap,
    nrow => :n_regions
)
sort!(country_gap, :mean_gap)

println("Largest gender gaps (F - M, negative = women worse off):")
first(country_gap, 15)
```

```julia
println("Smallest gender gaps (most equal or women ahead):")
last(sort(country_gap, :mean_gap), 15)
```
